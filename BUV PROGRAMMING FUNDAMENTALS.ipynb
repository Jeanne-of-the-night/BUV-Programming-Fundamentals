## Program 1 - Operational Budgeting Manager

Establishes the business's baseline financial health. It tracks initial funding pool contributions, allocates monthly budget caps across categories (inventory, marketing, utilities), logs fixed operational costs, and issues warnings when category spending nears budget limits.

```python
budget = 1000
budget_cap = 1500

costs = {
    "storage": 200,
    "marketing": 300,
    "utilities": 150,
    "website": 100,
    "staff_salaries": 400
}

def check_budget():
    total_costs = sum(costs.values())
    if total_costs > budget_cap:
        return "Budget exceeded!"
    elif total_costs > budget:
        return "Warning: Approaching budget limit!"
    else:
        return "Within budget."

if __name__ == '__main__':
    print(check_budget())
```

## Program 6: Customer Loyalty & Tiered Rewards Program

Manages customer retention. It maintains a database of registered student/faculty members, awards points per dollar spent, tracks loyalty tiers (Bronze, Silver, Gold), and calculates redeemable cash-back or reward coupons during checkout.

```python
# DATA STORAGE 
customers_data = [
    {"id": "C000001", "name": "Aemeathreiko", "points": 99, "tier": "Bronze"},
    {"id": "C000002", "name": "Bouncy", "points": 100, "tier": "Silver"},
    {"id": "C000003", "name": "Creepy", "points": 300, "tier": "Gold"},
]


# Functions for Customer Management, Loyalty and Tiered
def tier_check(points):
    """Checks the tier of a customer based on their points. Using if-elif-else."""
    if points >= 300:
        return "Gold"
    elif points >= 100:
        return "Silver"
    else:
        return "Bronze"


def customer_view():
    """Displays the list of all customers. Using for loop."""
    print("Customer List:")
    for customer in customers_data:
        print(f"ID: {customer['id']}, Name: {customer['name']}, Points: {customer['points']}, Tier: {customer['tier']}")


def customer_register():
    """Registers a new customer. Using input() and appending to the list."""
    new_id = input("Enter new customer ID (e.g., C110307): ")
    for customer in customers_data:
        if customer['id'] == new_id:
            print("Customer ID already exists. Registration failed.")
            return

    new_name = input("Enter customer name: ")
    new_customer = {"id": new_id, "name": new_name, "points": 0, "tier": "Bronze"}
    customers_data.append(new_customer)
    print(f"Customer {new_name} registered successfully.")


def add_points():
    """Adds points to a customer's account. Using input() and updating the list."""
    customer_id = input("Enter customer ID to add points: ")

    for customer in customers_data:
        if customer['id'] == customer_id:
            amount = int(input("Enter points to add (1 point per $1 spent, rounded down): "))
            points_earned = int(amount)
            customer['points'] += points_earned
            customer['tier'] = tier_check(customer['points'])
            print(f"Added {points_earned} points to {customer['name']}. New points: {customer['points']}, New tier: {customer['tier']}")
            return

    print("Customer ID not found. Points addition failed.")


def points_redeem():
    """Redeems points for a customer. Using input() and updating the list."""
    customer_id = input("Enter customer ID to redeem points: ")

    for customer in customers_data:
        if customer['id'] == customer_id:
            points_to_redeem = int(input("Enter points to redeem: "))
            if points_to_redeem > customer['points']:
                print(f"Insufficient points. {customer['name']} has only {customer['points']} points.")
                return
            customer['points'] -= points_to_redeem
            customer['tier'] = tier_check(customer['points'])
            print(f"Redeemed {points_to_redeem} points from {customer['name']}. New points: {customer['points']}, New tier: {customer['tier']}")
            return

    print("Customer ID not found. Points redemption failed.")


# Main Menu

def main_menu():
    """Displays the main menu and handles user input. Using while loop and if-elif-else."""
    while True:
        print("\n--- Main Menu ---")
        print("1. View Customers")
        print("2. Register Customer")
        print("3. Add Points")
        print("4. Redeem Points")
        print("5. Exit")
        choice = input("Enter your choice (1-5): ")

        if choice == '1':
            customer_view()
        elif choice == '2':
            customer_register()
        elif choice == '3':
            add_points()
        elif choice == '4':
            points_redeem()
        elif choice == '5':
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    main_menu()
```

## Program 7 - Vendor Purchase Order System

Handles upstream supply chain tasks. It maintains a directory of wholesale suppliers, drafts formal purchase orders for raw materials, tracks order statuses (Pending, Delivered, Paid), and calculates the total cost of goods ordered.

```python
import sys
vendors = {
    "V001": {
        "Coffee": 3.50,
        "Tea": 2.50,
        "Sandwich": 5.00
    },
    "V002": {
        "Salad": 4.00,
        "Muffin": 2.00,
        "Cookie": 1.50
    },
    "V003": {
        "Cheesecake": 4.50,
        "Smoothie": 3.75,
        "Juice": 3.00
    }
}

def check():
    vendor_id = input("Vendor ID: ").upper()
    
    if not vendor_id:
        print("❌ Error: Vendor ID cannot be empty.")
        sys.exit()
    if vendor_id not in vendors:
        print("❌ Error: Vendor ID does not exist.")
        sys.exit()
    
    item_name = input("Item Name: ").capitalize()
    
    if not item_name:
        print("❌ Error: Item Name cannot be empty.")
        sys.exit()

    if item_name not in vendors[vendor_id]:
        print("❌ Error: Vendor does not sell the item.")
        sys.exit()

    purchase_quantity = input("Purchase Quantity: ").strip()

    if not purchase_quantity:
        print("❌ Error: Purchase quantity cannot be empty.")
        sys.exit()
    if not purchase_quantity.isdigit():
        print("❌ Error: Purchase quantity must be a positive whole number.")
        sys.exit()
    if int(purchase_quantity) <= 0:
        print("❌ Error: Purchase quantity must be at least 1.")
        sys.exit()

    if int(purchase_quantity) > 100:
        print("❌ Error: Purchase quantity cannot exceed 100.")
        sys.exit()

    statuses = ["OUT", "PENDING", "IN STOCK"]
    print("1. OUT\n2. PENDING\n3. IN STOCK")
    choice = input("Status: ")
    if choice == "1":
        status = "OUT"
    elif choice == "2":
        status = "PENDING"
    elif choice == "3":
        status = "IN STOCK"
    else:
        print("❌ Error: Invalid status")
        sys.exit()
    unit_price = vendors[vendor_id][item_name]

    total_cost = int(purchase_quantity) * unit_price

    print("\n===== PURCHASE ORDER =====")
    print(f"Vendor ID:           {vendor_id}")
    print(f"Item Name:           {item_name}")
    print(f"Quantity:            {purchase_quantity}")
    print(f"Unit Price:          ${unit_price:.2f}")
    print(f"Total Cost of Goods: ${total_cost:.2f}")
    print(f"Status:              {status}")

def main():
    check()

try:
    if __name__ == '__main__':
        main()
except SystemExit:
    pass
```

## Program 8 - Business Analytics Report

Provides actionable operational insights. It analyzes transaction logs to rank top-selling products, identify peak revenue hours, calculate average order values, and highlight low-performing inventory items.

```python
products = [
    ("P001", "Coffee", 3.50, 40, 0.10),
    ("P002", "Tea", 2.50, 25, 0.05),
    ("P003", "Sandwich", 5.00, 15, 0.15),
    ("P004", "Muffin", 2.00, 60, 0.00),
    ("P005", "Smoothie", 3.75, 10, 0.20),
]

def sort_by_price(data):
    """Sorts products by Base Price in ascending order."""
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j][2] > data[j+1][2]:
                data[j], data[j+1] = data[j+1], data[j]
    return data

def calculate_promo_metrics(base_price, stock, discount_rate):
    """Calculates discounted price, total revenue, and projected savings."""
    promo_price = base_price * (1 - discount_rate)
    projected_revenue = promo_price * stock
    total_savings = (base_price - promo_price) * stock
    return promo_price, projected_revenue, total_savings

def generate_business_report(data):
    """Prints the analytical sales and promotional report."""
    print("=" * 70)
    print("             PROGRAM 8: BUSINESS ANALYTICS REPORT")
    print("=" * 70)
    
    print(f"{'ID':<6} | {'Item Name':<12} | {'Base Price':<10} | {'Promo Price':<11} | {'Proj. Revenue':<14}")
    print("-" * 70)
    
    total_revenue = 0
    total_customer_savings = 0
    
    for item in data:
        code, name, base_price, stock, discount_rate = item
        promo_price, revenue, savings = calculate_promo_metrics(base_price, stock, discount_rate)
        
        total_revenue += revenue
        total_customer_savings += savings
        
        print(f"{code:<6} | {name:<12} | ${base_price:<9.2f} | ${promo_price:<10.2f} | ${revenue:<13.2f}")
    
    print("-" * 70)
    print(f"Total Projected Revenue : ${total_revenue:.2f}")
    print(f"Total Promo Discounts   : ${total_customer_savings:.2f}")
    print("=" * 70)


def main():
    print("Sorting products by base price...")
    sorted_products = sort_by_price(products)
    generate_business_report(sorted_products)

if __name__ == '__main__':
    main()
```

## Program 9: Multi-Currency & Cross-Border Event Converter

Multi-Currency & Cross-Border Event Converter
Description: Accommodates international events, pop-ups, or imported inventory. It updates local exchange rates, converts foreign material costs into local currency, and calculates exchange fees for cross-border pop-up markets.

```python
#Functions for Currency Conversion
# Exchange rates (base on 1 USD). Data taken in 24/08/2026
rates = {
    "USD": 1.0,
    "EUR": 0.86,
    "GBP": 0.73,
    "JPY": 158.89,
    "AUD": 1.40,
    "CNY": 6.72,
    "VND": 26172.0
} 

def view_exchange_rates():
    """Displays the current exchange rates. Using for loop."""
    print("Current Exchange Rates (base on 1 USD):")
    for currency in rates:
        print(f"{currency}: {rates[currency]}")


def currency_convert():
    """Converts an amount from one currency to another. Using input() and calculations."""
    amount = float(input("Enter amount to convert: "))
    from_currency = input("Enter the currency to convert from (only available for USD, EUR, GBP, JPY, AUD, CNY, VND): ").upper()
    to_currency = input("Enter the currency to convert to (only available for USD, EUR, GBP, JPY, AUD, CNY, VND): ").upper()

    if amount < 0:
        print("Amount cannot be negative. Conversion failed.")
        return
    
    if from_currency in rates and to_currency in rates:
        usd_amount = amount / rates[from_currency]  # Convert to USD first
        converted_amount = usd_amount * rates[to_currency]  # Then convert to target currency
        print(f"{amount} {from_currency} is equal to {converted_amount:.2f} {to_currency}.")
    else:
        print("Invalid currency code. Conversion failed.")


def cross_border_fee():
    """Calculates the cross-border fee for a transaction. Using input() and calculations."""
    amount = float(input("Enter transaction amount in USD: "))
    if amount < 0:
        print("Amount cannot be negative. Fee calculation failed.")
        return
    fee_percentage = 0.02  # 2% fee
    fee = amount * fee_percentage
    total_amount = amount + fee
    print(f"Transaction Amount: ${amount:.2f}, Cross-Border Fee: ${fee:.2f}, Total Amount: ${total_amount:.2f}")

def main_menu():
    """Displays the main menu and handles user input. Using while loop and if-elif-else."""
    while True:
        print("1. View Exchange Rates")
        print("2. Currency Conversion")
        print("3. Cross-Border Fee Calculation")
        print("4. Exit")
        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            view_exchange_rates()
        elif choice == '2':
            currency_convert()
        elif choice == '3':
            cross_border_fee()
        elif choice == '4':
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    main_menu()
```

## Program 10 - Dynamic Pricing Simulation

Runs promotional experiments before launching them live. It models "buy-one-get-one" (BOGO) offers, tiered volume discounts, time-based flash sales, and bundle pricing to project profit margins under different pricing strategies.

```python

products = [
    ("P001", "Coffee", 3.50, 8, 9),
    ("P002", "Tea", 2.50, 30, 2),
    ("P003", "Sandwich", 5.00, 15, 5),
    ("P004", "Muffin", 2.00, 25, 3),
    ("P005", "Smoothie", 3.75, 5, 8),
]

def calculate_dynamic_price(base_price, stock, demand):
    """
    Adjusts price based on stock and demand:
    - High Demand (>= 7) & Low Stock (< 10)  -> Surge +20%
    - Low Demand (<= 4)  & High Stock (>= 20) -> Discount -15%
    - Normal conditions                      -> Standard price
    """
    if demand >= 7 and stock < 10:
        multiplier = 1.20
        status = "SURGE (+20%)"
    elif demand <= 4 and stock >= 20:
        multiplier = 0.85
        status = "DISCOUNT (-15%)"
    else:
        multiplier = 1.00
        status = "STANDARD"
        
    dynamic_price = base_price * multiplier
    return dynamic_price, status

def run_pricing_simulation(data):
    """Runs simulation and displays dynamic price adjustments."""
    print("=" * 72)
    print("          PROGRAM 10: DYNAMIC PRICING SIMULATION")
    print("=" * 72)
    
    print(f"{'ID':<6} | {'Item Name':<10} | {'Base Price':<10} | {'Dyn. Price':<10} | {'Pricing Rule':<16}")
    print("-" * 72)
    
    total_original_val = 0
    total_dynamic_val = 0
    
    for item in data:
        code, name, base_price, stock, demand = item
        dyn_price, status = calculate_dynamic_price(base_price, stock, demand)
        
        total_original_val += base_price * stock
        total_dynamic_val += dyn_price * stock
        
        print(f"{code:<6} | {name:<10} | ${base_price:<9.2f} | ${dyn_price:<9.2f} | {status:<16}")
    
    variance = total_dynamic_val - total_original_val
    
    print("-" * 72)
    print(f"Original Inventory Value : ${total_original_val:.2f}")
    print(f"Dynamic Inventory Value  : ${total_dynamic_val:.2f}")
    print(f"Total Revenue Impact     : ${variance:+.2f}")
    print("=" * 72)

def main():
    print("Running Dynamic Pricing Engine...")
    run_pricing_simulation(products)

if __name__ == '__main__':
    main()
```

```python
import sys

# ==========================================
# PROGRAM 1: BUDGET TRACKER
# ==========================================
p1_budget = 1000
p1_budget_cap = 1500

p1_costs = {
    "storage": 200,
    "marketing": 300,
    "utilities": 150,
    "website": 100,
    "staff_salaries": 400
}

def p1_check_budget():
    total_costs = sum(p1_costs.values())
    print("\n--- Program 1: Budget Tracker ---")
    print(f"Total Costs: ${total_costs}")
    if total_costs > p1_budget_cap:
        return "Budget exceeded!"
    elif total_costs > p1_budget:
        return "Warning: Approaching budget limit!"
    else:
        return "Within budget."

def run_program_1():
    print(p1_check_budget())


# ==========================================
# PROGRAM 2: INVENTORY THRESHOLD CHECKER
# ==========================================
p2_menu = {
    "Coffee": {"item_id": "ITEM101", "stock": 3, "threshold": 5},
    "Tea": {"item_id": "ITEM102", "stock": 10, "threshold": 5},
    "Sandwich": {"item_id": "ITEM103", "stock": 2, "threshold": 5}
}

def p2_search():
    item = input("Item Name (e.g. Coffee, Tea, Sandwich): ").strip().capitalize()
    if not item:
        print("❌ Error: Item name cannot be empty.")
        return None
    if item not in p2_menu:
        print(f"❌ Error: '{item}' was not found in the menu inventory.")
        return None
    return item

def p2_show_results(item):   
    item_id = p2_menu[item]["item_id"]
    stock = p2_menu[item]["stock"]
    threshold = p2_menu[item]["threshold"]
    status = "REORDER!" if stock < threshold else "OK"

    print("\n===== INVENTORY CARD =====")
    print(f"Item ID:    {item_id}")
    print(f"Item Name:  {item}")
    print(f"Stock:      {stock}")
    print(f"Threshold:  {threshold}")
    print(f"Status:     {status}")

def run_program_2():
    print("\n--- Program 2: Inventory Checker ---")
    item = p2_search()
    if item:
        p2_show_results(item)


# ==========================================
# PROGRAM 3: CAMPUS CAFE POS TERMINAL
# ==========================================
p3_menu = {
    "Coffee": 3.50,
    "Tea": 2.50,
    "Sandwich": 5.00,
    "Salad": 4.00,
    "Muffin": 2.00,
    "Cookie": 1.50,
    "Cheesecake": 4.50,
    "Smoothie": 3.75,
    "Juice": 3.00,
    "Espresso": 2.75,
    "Green tea": 2.45,
}

p3_promo_codes = {
    "DISCOUNT10": 0.10,  # 10% discount
    "STUDENT90": 0.20,   # 20% discount
}

def p3_order():
    order_input = input("Enter items to order (Exp: Coffee 2, Tea 1, Sandwich 1): ")
    print("Order received. Processing your order...")
    promo_code = input("Enter a promo code (if applicable): ").strip().upper()

    if promo_code and promo_code not in p3_promo_codes:
        print(f"Warning: '{promo_code}' is an invalid promo code. Continuing without discount.\n")

    order_list = []
    for item in order_input.split(","):
        parts = item.strip().split()
        if len(parts) < 2:
            continue
        name = " ".join(parts[:-1]).capitalize()
        qty = parts[-1]

        if name in p3_menu and qty.isdigit() and int(qty) > 0:
            order_list.append({name: int(qty)})

    if not order_list:
        print("No valid items were entered. Please enter at least one valid item from the menu to proceed.")
        return None, None

    return order_list, promo_code

def p3_apply_discounts(sub_total, promo_code, promo_codes):
    if promo_code in promo_codes:
        discount_rate = promo_codes[promo_code]
        return sub_total * discount_rate
    return 0.0

def p3_calculate_price(menu, item, quantity):
    if item in menu:
        return menu[item] * quantity
    return 0.0  

def p3_calculate_total(menu, order_list):
    total_price = 0.0
    for order_item in order_list:
        for item, quantity in order_item.items():
            total_price += p3_calculate_price(menu, item, quantity)
    return total_price

def p3_bills(menu, promo_codes, order_list, promo_code=""):
    sub_total = p3_calculate_total(menu, order_list)
    discount_amount = p3_apply_discounts(sub_total, promo_code, promo_codes)
    tax = sub_total * 0.05
    total_due = sub_total - discount_amount + tax
    
    print("\n" + "=" * 46)
    print("         CAMPUS CAFE POS TERMINAL             ")
    print("=" * 46)
    print(f"{'Item':<18} {'Qty':^5} {'Price':>9} {'Total':>10}")
    print("-" * 46)

    for i, item_dict in enumerate(order_list, start=1):
        for item, qty in item_dict.items():
            total = p3_calculate_price(menu, item, qty)
            item_label = f"{i}. {item}"
            print(f"{item_label:<18} {qty:^5} {f'${menu[item]:.2f}':>9} {f'${total:.2f}':>10}")

    print("-" * 48)
    print(f"{'Subtotal:':<33} {f'${sub_total:.2f}':>10}")

    if discount_amount > 0:
        print(f"{'Discount:':<33} {f'-${discount_amount:.2f}':>10}")
    print(f"{'Tax (5%):':<33} {f'${tax:.2f}':>10}")
    print("=" * 46)
    print("-" * 48)
    print(f"{'TOTAL DUE:':<33} {f'${total_due:.2f}':>10}")
    print("=" * 46)
    print(" [1] Complete Sale | [2] Cancel Order")

def run_program_3():
    print("\n--- Program 3: POS Terminal ---")
    order_list, promo_code = p3_order()
    if order_list:
        p3_bills(p3_menu, p3_promo_codes, order_list, promo_code)


# ==========================================
# PROGRAM 4: EMPLOYEE SCHEDULING & PAYROLL
# ==========================================
p4_staff_name = "Alex Smith"
p4_staff_id = "STU-8821"
p4_hours_worked = 15.0 
p4_hourly_rate = 12.00
p4_overtime_hours = 2.0
p4_overtime_rate = 18.00

def p4_calculate_payroll():
    gross_pay = (p4_hours_worked * p4_hourly_rate) + (p4_overtime_hours * p4_overtime_rate)
    tax_reserve = gross_pay * 0.10
    net_payout = gross_pay - tax_reserve
    return gross_pay, tax_reserve, net_payout 

def run_program_4():
    gross_pay, tax_reserve, net_payout = p4_calculate_payroll()
    print("\nProg 4: Employee Scheduling & Payroll Calculator")
    print("=" * 40)
    print("STUDENT PAYSLIP")
    print("=" * 40)
    print(f"Staff Name: {p4_staff_name} (ID: {p4_staff_id})")
    print(f"Hours Worked: {p4_hours_worked} hrs @ ${p4_hourly_rate:.2f}/hr")
    print(f"Overtime Hours: {p4_overtime_hours} hrs @ ${p4_overtime_rate:.2f}/hr")
    print("=" * 40)
    print(f"Gross Pay: ${gross_pay:.2f}")
    print(f"Tax Reserve (10%): -${tax_reserve:.2f}")
    print("=" * 40)
    print(f"NET PAYOUT: ${net_payout:.2f}")
    print("=" * 40)


# ==========================================
# PROGRAM 5: FINANCIAL PERFORMANCE SUMMARY
# ==========================================
p5_revenue_p3 = 9.23
p5_cogs_p2_p7 = 150.00
p5_expense_p1 = 25.00
p5_payroll_p4 = 216.00
p5_expenses_p1_p4 = p5_expense_p1 + p5_payroll_p4

def p5_calculate_financials():
    gross_profit = p5_revenue_p3 - p5_cogs_p2_p7
    net_profit = gross_profit - p5_expenses_p1_p4
    return gross_profit, net_profit

def run_program_5():
    gross_profit, net_profit = p5_calculate_financials()
    print("\nProg 5: Financial Performance & Profit/Loss Estimator")
    print("=" * 40)
    print("FINANCIAL PERFORMANCE SUMMARY")
    print("=" * 40)
    print(f"Revenue (Prog 3): ${p5_revenue_p3:.2f}")
    print(f"COGS (Prog 2 + 7): -${p5_cogs_p2_p7:.2f}")
    print("-" * 40)
    print(f"Gross Profit: ${gross_profit:.2f}")
    print(f"Expenses (Prog 1 + 4): -${p5_expenses_p1_p4:.2f}")
    print("-" * 40)
    print(f"NET PROFIT/LOSS: ${net_profit:.2f}")
    print("=" * 40)


# ==========================================
# PROGRAM 6: CUSTOMER LOYALTY MANAGEMENT
# ==========================================
p6_customers_data = [
    {"id": "C000001", "name": "Aemeathreiko", "points": 99, "tier": "Bronze"},
    {"id": "C000002", "name": "Bouncy", "points": 100, "tier": "Silver"},
    {"id": "C000003", "name": "Creepy", "points": 300, "tier": "Gold"},
]

def p6_tier_check(points):
    if points >= 300:
        return "Gold"
    elif points >= 100:
        return "Silver"
    else:
        return "Bronze"

def p6_customer_view():
    print("\nCustomer List:")
    for customer in p6_customers_data:
        print(f"ID: {customer['id']}, Name: {customer['name']}, Points: {customer['points']}, Tier: {customer['tier']}")

def p6_customer_register():
    new_id = input("Enter new customer ID (e.g., C110307): ")
    for customer in p6_customers_data:
        if customer['id'] == new_id:
            print("Customer ID already exists. Registration failed.")
            return

    new_name = input("Enter customer name: ")
    new_customer = {"id": new_id, "name": new_name, "points": 0, "tier": "Bronze"}
    p6_customers_data.append(new_customer)
    print(f"Customer {new_name} registered successfully.")

def p6_add_points():
    customer_id = input("Enter customer ID to add points: ")
    for customer in p6_customers_data:
        if customer['id'] == customer_id:
            amount = int(input("Enter points to add: "))
            customer['points'] += amount
            customer['tier'] = p6_tier_check(customer['points'])
            print(f"Added {amount} points to {customer['name']}. New points: {customer['points']}, New tier: {customer['tier']}")
            return
    print("Customer ID not found. Points addition failed.")

def p6_points_redeem():
    customer_id = input("Enter customer ID to redeem points: ")
    for customer in p6_customers_data:
        if customer['id'] == customer_id:
            points_to_redeem = int(input("Enter points to redeem: "))
            if points_to_redeem > customer['points']:
                print(f"Insufficient points. {customer['name']} has only {customer['points']} points.")
                return
            customer['points'] -= points_to_redeem
            customer['tier'] = p6_tier_check(customer['points'])
            print(f"Redeemed {points_to_redeem} points from {customer['name']}. New points: {customer['points']}, New tier: {customer['tier']}")
            return
    print("Customer ID not found. Points redemption failed.")

def run_program_6():
    while True:
        print("\n--- Program 6: Loyalty Management Menu ---")
        print("1. View Customers")
        print("2. Register Customer")
        print("3. Add Points")
        print("4. Redeem Points")
        print("5. Back to Main Application Menu")
        choice = input("Enter choice (1-5): ")

        if choice == '1':
            p6_customer_view()
        elif choice == '2':
            p6_customer_register()
        elif choice == '3':
            p6_add_points()
        elif choice == '4':
            p6_points_redeem()
        elif choice == '5':
            break
        else:
            print("Invalid choice. Please try again.")


# ==========================================
# PROGRAM 7: VENDOR PURCHASE ORDER
# ==========================================
p7_vendors = {
    "V001": {"Coffee": 3.50, "Tea": 2.50, "Sandwich": 5.00},
    "V002": {"Salad": 4.00, "Muffin": 2.00, "Cookie": 1.50},
    "V003": {"Cheesecake": 4.50, "Smoothie": 3.75, "Juice": 3.00}
}

def run_program_7():
    print("\n--- Program 7: Vendor Purchase Order ---")
    vendor_id = input("Vendor ID (V001, V002, V003): ").upper()
    if not vendor_id or vendor_id not in p7_vendors:
        print("❌ Error: Invalid or missing Vendor ID.")
        return

    item_name = input("Item Name: ").capitalize()
    if not item_name or item_name not in p7_vendors[vendor_id]:
        print("❌ Error: Invalid or unhandled Item Name for this vendor.")
        return

    purchase_quantity = input("Purchase Quantity: ").strip()
    if not purchase_quantity.isdigit() or int(purchase_quantity) <= 0 or int(purchase_quantity) > 100:
        print("❌ Error: Purchase quantity must be a positive integer between 1 and 100.")
        return

    print("Status Options:\n1. OUT\n2. PENDING\n3. IN STOCK")
    choice = input("Status Choice: ")
    status_map = {"1": "OUT", "2": "PENDING", "3": "IN STOCK"}
    if choice not in status_map:
        print("❌ Error: Invalid status selection.")
        return

    status = status_map[choice]
    unit_price = p7_vendors[vendor_id][item_name]
    total_cost = int(purchase_quantity) * unit_price

    print("\n===== PURCHASE ORDER =====")
    print(f"Vendor ID:           {vendor_id}")
    print(f"Item Name:           {item_name}")
    print(f"Quantity:            {purchase_quantity}")
    print(f"Unit Price:          ${unit_price:.2f}")
    print(f"Total Cost of Goods: ${total_cost:.2f}")
    print(f"Status:              {status}")


# ==========================================
# PROGRAM 8: BUSINESS ANALYTICS REPORT
# ==========================================
p8_products = [
    ("P001", "Coffee", 3.50, 40, 0.10),
    ("P002", "Tea", 2.50, 25, 0.05),
    ("P003", "Sandwich", 5.00, 15, 0.15),
    ("P004", "Muffin", 2.00, 60, 0.00),
    ("P005", "Smoothie", 3.75, 10, 0.20),
]

def p8_sort_by_price(data):
    data_copy = list(data)
    n = len(data_copy)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data_copy[j][2] > data_copy[j+1][2]:
                data_copy[j], data_copy[j+1] = data_copy[j+1], data_copy[j]
    return data_copy

def p8_calculate_promo_metrics(base_price, stock, discount_rate):
    promo_price = base_price * (1 - discount_rate)
    projected_revenue = promo_price * stock
    total_savings = (base_price - promo_price) * stock
    return promo_price, projected_revenue, total_savings

def run_program_8():
    print("\nSorting products by base price...")
    sorted_products = p8_sort_by_price(p8_products)
    
    print("=" * 70)
    print("              PROGRAM 8: BUSINESS ANALYTICS REPORT")
    print("=" * 70)
    print(f"{'ID':<6} | {'Item Name':<12} | {'Base Price':<10} | {'Promo Price':<11} | {'Proj. Revenue':<14}")
    print("-" * 70)
    
    total_revenue = 0
    total_customer_savings = 0
    
    for item in sorted_products:
        code, name, base_price, stock, discount_rate = item
        promo_price, revenue, savings = p8_calculate_promo_metrics(base_price, stock, discount_rate)
        total_revenue += revenue
        total_customer_savings += savings
        print(f"{code:<6} | {name:<12} | ${base_price:<9.2f} | ${promo_price:<10.2f} | ${revenue:<13.2f}")
    
    print("-" * 70)
    print(f"Total Projected Revenue : ${total_revenue:.2f}")
    print(f"Total Promo Discounts   : ${total_customer_savings:.2f}")
    print("=" * 70)


# ==========================================
# PROGRAM 9: CURRENCY CONVERTER
# ==========================================
p9_rates = {
    "USD": 1.0,
    "EUR": 0.86,
    "GBP": 0.73,
    "JPY": 158.89,
    "AUD": 1.40,
    "CNY": 6.72,
    "VND": 26172.0
} 

def p9_view_exchange_rates():
    print("\nCurrent Exchange Rates (base on 1 USD):")
    for currency, rate in p9_rates.items():
        print(f"{currency}: {rate}")

def p9_currency_convert():
    try:
        amount = float(input("Enter amount to convert: "))
    except ValueError:
        print("Invalid number input.")
        return

    from_currency = input("Convert from (USD, EUR, GBP, JPY, AUD, CNY, VND): ").upper()
    to_currency = input("Convert to (USD, EUR, GBP, JPY, AUD, CNY, VND): ").upper()

    if amount < 0:
        print("Amount cannot be negative. Conversion failed.")
        return
    
    if from_currency in p9_rates and to_currency in p9_rates:
        usd_amount = amount / p9_rates[from_currency]
        converted_amount = usd_amount * p9_rates[to_currency]
        print(f"{amount:.2f} {from_currency} = {converted_amount:.2f} {to_currency}.")
    else:
        print("Invalid currency code. Conversion failed.")

def p9_cross_border_fee():
    try:
        amount = float(input("Enter transaction amount in USD: "))
    except ValueError:
        print("Invalid number input.")
        return

    if amount < 0:
        print("Amount cannot be negative. Fee calculation failed.")
        return

    fee = amount * 0.02  # 2% fee
    total_amount = amount + fee
    print(f"Transaction Amount: ${amount:.2f}, Cross-Border Fee (2%): ${fee:.2f}, Total Amount: ${total_amount:.2f}")

def run_program_9():
    while True:
        print("\n--- Program 9: Currency Converter Menu ---")
        print("1. View Exchange Rates")
        print("2. Currency Conversion")
        print("3. Cross-Border Fee Calculation")
        print("4. Back to Main Application Menu")
        choice = input("Enter choice (1-4): ")

        if choice == '1':
            p9_view_exchange_rates()
        elif choice == '2':
            p9_currency_convert()
        elif choice == '3':
            p9_cross_border_fee()
        elif choice == '4':
            break
        else:
            print("Invalid choice. Please try again.")


# ==========================================
# PROGRAM 10: DYNAMIC PRICING SIMULATION
# ==========================================
p10_products = [
    ("P001", "Coffee", 3.50, 8, 9),
    ("P002", "Tea", 2.50, 30, 2),
    ("P003", "Sandwich", 5.00, 15, 5),
    ("P004", "Muffin", 2.00, 25, 3),
    ("P005", "Smoothie", 3.75, 5, 8),
]

def p10_calculate_dynamic_price(base_price, stock, demand):
    if demand >= 7 and stock < 10:
        multiplier = 1.20
        status = "SURGE (+20%)"
    elif demand <= 4 and stock >= 20:
        multiplier = 0.85
        status = "DISCOUNT (-15%)"
    else:
        multiplier = 1.00
        status = "STANDARD"
        
    dynamic_price = base_price * multiplier
    return dynamic_price, status

def run_program_10():
    print("\nRunning Dynamic Pricing Engine...")
    print("=" * 72)
    print("          PROGRAM 10: DYNAMIC PRICING SIMULATION")
    print("=" * 72)
    print(f"{'ID':<6} | {'Item Name':<10} | {'Base Price':<10} | {'Dyn. Price':<10} | {'Pricing Rule':<16}")
    print("-" * 72)
    
    total_original_val = 0
    total_dynamic_val = 0
    
    for item in p10_products:
        code, name, base_price, stock, demand = item
        dyn_price, status = p10_calculate_dynamic_price(base_price, stock, demand)
        
        total_original_val += base_price * stock
        total_dynamic_val += dyn_price * stock
        
        print(f"{code:<6} | {name:<10} | ${base_price:<9.2f} | ${dyn_price:<9.2f} | {status:<16}")
    
    variance = total_dynamic_val - total_original_val
    
    print("-" * 72)
    print(f"Original Inventory Value : ${total_original_val:.2f}")
    print(f"Dynamic Inventory Value  : ${total_dynamic_val:.2f}")
    print(f"Total Revenue Impact     : ${variance:+.2f}")
    print("=" * 72)


# ==========================================
# MASTER APPLICATION CONTROLLER
# ==========================================
def main_menu():
    while True:
        print("\n==========================================")
        print("      CAMPUS MANAGEMENT SYSTEM SUITE     ")
        print("==========================================")
        print(" 1. Budget Tracker")
        print(" 2. Inventory Threshold Checker")
        print(" 3. Campus Cafe POS Terminal")
        print(" 4. Employee Payroll Calculator")
        print(" 5. Financial Performance Summary")
        print(" 6. Customer Loyalty Management")
        print(" 7. Vendor Purchase Order")
        print(" 8. Business Analytics Report")
        print(" 9. Currency Conversion System")
        print("10. Dynamic Pricing Simulation")
        print(" 0. Exit System")
        print("==========================================")
        
        choice = input("Select a program to execute (0-10): ").strip()

        if choice == '1':
            run_program_1()
        elif choice == '2':
            run_program_2()
        elif choice == '3':
            run_program_3()
        elif choice == '4':
            run_program_4()
        elif choice == '5':
            run_program_5()
        elif choice == '6':
            run_program_6()
        elif choice == '7':
            run_program_7()
        elif choice == '8':
            run_program_8()
        elif choice == '9':
            run_program_9()
        elif choice == '10':
            run_program_10()
        elif choice == '0':
            print("Exiting suite. Goodbye!")
            sys.exit()
        else:
            print("Invalid input. Please choose a valid option (0-10).")

if __name__ == '__main__':
    main_menu()
```
